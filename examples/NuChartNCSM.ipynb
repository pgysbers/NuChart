```python
import sys
sys.path.append("../../")
from NuChart.src.NuChart import *
import pandas as pd
import numpy as np
```

```python
nc = NuChart(55, 45, figsize=(14, 8))
nc.add_magic_numbers()

def plot_HT(nc):
    candidates_HT15_parent = [(6,10), (8,14), (12,22), (13,26), (14,26), (17,34), (18,34), (19,38), (20,38), (21,42), (23,46), (25,50), (27,54), (31,62), (37,74) ] 
    candidates_HT15_daught = [(5,10), (7,14), (11,22), (12,26), (13,26), (16,34), (17,34), (18,38), (19,38), (20,42), (22,46), (24,50), (26,54), (30,62), (36,74)] 
    candidates_HT23_parent = candidates_HT15_parent + [(10,18), (16,30), (22,42), (24,46), (26,50), (28,54), (33,66), (35,70)]
    candidates_HT23_daught = candidates_HT15_daught + [(9,18), (15,30), (21,42), (23,46), (25,50), (27,54), (32,66), (34,70)]

    N = Z = np.linspace(0.5,nc.ylim)
    nc.get_axis().plot(N,Z,'k--', label='$N=Z$')
    nc.get_axis().text(43,42,'$N=Z$')

    from matplotlib.markers import MarkerStyle

    mp = MarkerStyle(">", fillstyle='none')
    mp._transform.rotate_deg(-45)

    md = MarkerStyle("<", fillstyle='none')
    md._transform.rotate_deg(-45)
    nc.highlight_candidates(candidates_HT23_parent, marker=mp, color='tab:orange', legend=False)
    nc.highlight_candidates(candidates_HT23_daught, marker=md, color='tab:blue', legend=False)

    mp = MarkerStyle(">")
    mp._transform.rotate_deg(-45)

    md = MarkerStyle("<")
    md._transform.rotate_deg(-45)

    nc.highlight_candidates(candidates_HT15_parent, marker=mp, color='tab:orange', label='Parent', legend=False)
    nc.highlight_candidates(candidates_HT15_daught, marker=md, color='tab:blue', label='Daughter', legend=False)

    nc.add_candidate_legend(title='Super-Allowed Fermi Decays', loc='lower right')

plot_HT(nc)
```

```python
for Nmax in range(3):
    nc = NuChart(65,55, figsize=(14,8))
    nc.add_magic_numbers()
    nc.plot_ncsm(Nmax=Nmax)
    plot_HT(nc)
    nc.savefig(f'NCSMdim_Nmax{Nmax}.pdf')
```
